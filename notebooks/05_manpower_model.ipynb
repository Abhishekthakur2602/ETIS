```python
import pandas as pd

df = pd.read_csv(
    "../dataset/congestion_dataset.csv"
)
```

```python
def constables(row):

    score = row["congestion_score"]

    if score >= 80:
        return 8

    elif score >= 60:
        return 6

    elif score >= 40:
        return 4

    else:
        return 2
```

```python
def asi(row):

    score = row["congestion_score"]

    if score >= 80:
        return 2

    elif score >= 60:
        return 1

    return 0
```

```python
def inspector(row):

    if row["severity"] == "Critical":
        return 1

    return 0
```

```python
df["constables"] = df.apply(
    constables,
    axis=1
)

df["asi"] = df.apply(
    asi,
    axis=1
)

df["inspector"] = df.apply(
    inspector,
    axis=1
)
```
